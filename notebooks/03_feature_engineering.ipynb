```python
import pandas as pd 
df = pd.read_csv('../data/processed/train_with_target.csv', dtype={'StateHoliday': str})
df['Date'] = pd.to_datetime(df['Date']) 
df = df.sort_values(['Store', 'Date']).reset_index(drop=True) 
df.head() 
```

```python
df['SalesVelocity7'] = df.groupby('Store')['Sales'].transform(lambda x: x.shift(1).rolling(7, min_periods=1).mean()) 
df['SalesVelocity14'] = df.groupby('Store')['Sales'].transform(lambda x: x.shift(1).rolling(14, min_periods=1).mean())
```

```python
df['SalesTrend'] = df['SalesVelocity7'] - df['SalesVelocity14']
```

```python
df['DayOfWeek'] = df['Date'].dt.dayofweek
df['Month'] = df['Date'].dt.month
df['IsWeekend'] = df['DayOfWeek'].isin([5, 6]).astype(int)
```

```python
df.isnull().sum()
```

```python
df['SalesRatio'] = df['SalesRatio'].fillna(0) 
```

```python
df['SalesVelocity7'] = df['SalesVelocity7'].fillna(0) 
df['SalesVelocity14'] = df['SalesVelocity14'].fillna(0) 
df['SalesTrend'] = df['SalesTrend'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df.isnull().sum()
```

```python
df['DaysSinceLastVisit'] = df.groupby('Store')['Date'].diff().dt.days
df['DaysSinceLastVisit'] = df['DaysSinceLastVisit'].fillna(0) 
```

```python
df['CompetitionOpen'] = ( (df['CompetitionOpenSinceYear'] > 0) & ((df['Date'].dt.year > df['CompetitionOpenSinceYear']) | ((df['Date'].dt.year == df['CompetitionOpenSinceYear']) & (df['Date'].dt.month >= df['CompetitionOpenSinceMonth']))) ).astype(int)
```

```python
category_avg_stockout = df.groupby('StoreType')['LikelyStockout'].transform('mean') 
df['StoreTypeStockoutRate'] = category_avg_stockout
```

```python
df.isnull().sum()
```

```python
df.to_csv('../data/processed/train_features.csv', index=False)
```
